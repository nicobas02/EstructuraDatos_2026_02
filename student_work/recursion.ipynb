1. Escribir una funcion `potencia_dos(n)`que compute $2^n$
2. Una funcion recursiva `potencia_dos_recursiva(n)`que haga lo anterior
3. Una funcion `potencia_dos_recursiva1(n)`recursiva pero rapida que resuelva el problema anterior

```python
def potencia_dos(n):
  resultado = 1
  for i in range(n):
    resultado *= 2
  return resultado

potencia_dos(4),potencia_dos(0)
```

```python
def potencia_dos_recursiva(n):
  if n == 0:
    return 1
  else:
    return 2 * potencia_dos_recursiva(n-1)
```

```python
potencia_dos_recursiva(4)
```

```python
def potencia_dos_recursiva1(n):
  if n == 0:
    return 1
  if n % 2 == 0:
    half_pow = potencia_dos_recursiva1(n // 2)
    return half_pow * half_pow
  else:
    half_pow = potencia_dos_recursiva1((n - 1) // 2)
    return 2 * half_pow * half_pow
```

```python
potencia_dos_recursiva1(4)
```

```python
def maximo_recursivo(L):
  if len(L) == 1:
    return L[0]
  maximotentativo = maximo_recursivo(L[:-1])
  if maximotentativo > L[-1]:
    return maximotentativo
  else:
    return L[-1]
```
